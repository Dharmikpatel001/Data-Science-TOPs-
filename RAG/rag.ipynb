1. What is RAG?

RAG means Retrieval-Augmented Generation. It helps an AI find information from outside sources and then use it to answer a question.

Example: On Zomato, RAG could find the latest restaurant menu, price, and opening time before answering a customer.

2. Questions an LLM trained in 2022 may struggle with

Who won IPL 2024?

→ IPL 2024 happened after 2022.

Which teams reached the IPL 2024 playoffs?

→ This information was not available in 2022.

What is the latest IRCTC train schedule?

→ Train timings can change, so old information may be incorrect.

Problem: An old LLM may give outdated or wrong answers because it does not know newer information.

3. RAG Workflow

User Question
      ↓
Retrieve Documents
      ↓
Give Documents + Question to LLM
      ↓
Generate Answer

4. Comparison
|                    | **RAG**            | **Fine-tuning**     | **Prompt Engineering** |


| **Cost**           | Medium             | High                | Low                    |

| **Data Freshness** | High               | Low                 | Low                    |

| **Use Case**       | Zomato latest menu | Customer support AI | Asking ChatGPT clearly |

| **Speed**          | Medium             | Slow                | Very Fast              |


```python
def chunk_text(text, chunk_size, overlap):
    words = text.split()
    chunks = []
    start = 0

    while start < len(words):
        end = start + chunk_size
        chunks.append(" ".join(words[start:end]))
        start += chunk_size - overlap

    return chunks


# Small 500-word sample
sample_text = """
Wikipedia is a free online encyclopedia written and maintained by volunteers.
It contains millions of articles covering history, science, technology, geography,
culture, sports, and many other subjects. Wikipedia was launched in January 2001
and has become one of the most visited websites in the world. Anyone can read
Wikipedia articles, and many users can edit them. Editors add new information,
correct mistakes, improve references, and discuss changes. Wikipedia follows
rules about reliable sources and neutral points of view. Articles are organized
using links, categories, and references. The website is available in many languages.
Its open editing system allows information to be updated quickly, although readers
are encouraged to check important information using reliable sources. Wikipedia
is operated by the Wikimedia Foundation, a nonprofit organization. It also supports
other projects such as Wikimedia Commons, Wiktionary, and Wikibooks. Wikipedia
has played an important role in making general knowledge easily available on the
internet. Students, researchers, teachers, and everyday users often use it as a
starting point for learning about unfamiliar topics. The encyclopedia continues
to grow as volunteers around the world contribute knowledge and improve existing
articles. Its large collection of freely available information makes it a useful
resource for education and general research.
"""

chunks = chunk_text(sample_text, 100, 20)

print("Number of chunks:", len(chunks))
for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1}:\n{chunk}")
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer("all-MiniLM-L6-v2")

sentences = [
    "The burger was delicious",
    "I enjoyed the burger a lot"
]

embeddings = model.encode(sentences)

score = cosine_similarity(
    [embeddings[0]],
    [embeddings[1]]
)[0][0]

print("Cosine similarity:", score)
```

```python
import chromadb
import numpy as np

client = chromadb.Client()
collection = client.create_collection("restaurants")

names = [
    "Barbeque Nation",
    "Domino's",
    "McDonald's",
    "Pizza Hut",
    "Subway"
]

vectors = np.random.random((5, 3)).tolist()

collection.add(
    ids=["1", "2", "3", "4", "5"],
    embeddings=vectors,
    documents=names
)

query = np.random.random(3).tolist()

result = collection.query(
    query_embeddings=[query],
    n_results=1
)

print("Most similar restaurant:", result["documents"][0][0])
```

```python
import faiss
import numpy as np

movie_titles = [
    "Avengers",
    "Fast and Furious",
    "Titanic",
    "The Notebook",
    "John Wick"
]

# Random movie embeddings
movie_embeddings = np.random.random((5, 4)).astype("float32")

index = faiss.IndexFlatL2(4)
index.add(movie_embeddings)

def search_movies(query_embedding):
    query_embedding = np.array([query_embedding]).astype("float32")

    distances, positions = index.search(query_embedding, 2)

    print("Top 2 movies:")
    for pos in positions[0]:
        print(movie_titles[pos])

query = np.random.random(4)
search_movies(query)
```

```python
import PyPDF2

with open("whatsapp_chat.pdf", "rb") as file:
    reader = PyPDF2.PdfReader(file)

    text = ""
    for page in reader.pages:
        text += page.extract_text() or ""

chunks = []

for i in range(0, len(text), 500):
    chunks.append(text[i:i + 500])

print("Number of chunks:", len(chunks))
print(chunks[0])
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("paraphrase-MiniLM-L6-v2")

embeddings = model.encode(chunks)

print("Number of embeddings:", len(embeddings))
print("First embedding:", embeddings[0])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def find_relevant_chunk(question):
    question_embedding = model.encode([question])

    scores = cosine_similarity(
        question_embedding,
        embeddings
    )[0]

    best_index = scores.argmax()

    return chunks[best_index]

question = "Who messaged the most?"

answer = find_relevant_chunk(question)

print("Most relevant chunk:")
print(answer)
```
